```python
import torch
from diffusers import StableDiffusionPipeline
import os

# Load the Stable Diffusion model
pipeline = StableDiffusionPipeline.from_pretrained("runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16)
pipeline.to("cuda")
```

```python
prompt = "First dinasourus in wild nature" # @param {type:"string"}

# Generate the image
# You can adjust num_inference_steps for quality vs. speed tradeoff
image = pipeline(prompt, num_inference_steps=50).images[0]

# Create the /outputs directory if it doesn't exist
output_dir = "./outputs"
os.makedirs(output_dir, exist_ok=True)

# Define the output path for the image
image_path = os.path.join(output_dir, "generated_image.png")

image.save(image_path)

print(f"Image saved to {image_path}")
image # Display the image directly in the notebook
```
